```python
def ViewRecords(file_path):
  df = pd.read_excel(file_path) #Read the Excel file into a DataFrame
  print(df)
  ch = str(input("Press E to exit: "))
  if (ch=='e' or ch=='E'):
    clear_output(wait=True)
    Manager()
```

```python
def AddRecords():
  file_path = 'Manager.xlsx'
  wb = load_workbook(file_path)
  ws = wb.active
  data_to_add = []
  n1 = int(input("Enter Sr.No: "))
  n2 = str(input("Enter ID.No: "))
  n3 = str(input("Enter Name: "))
  n4 = str(input("Enter Profession: "))
  n5 = str(input("Enter Email Address: "))
  n6 =int(input("Enter Phone Number: "))
  n7 = int(input("Enter Salary: "))
  data_to_add.append([n1,n2,n3,n4,n5,n6,n7])
  for row in data_to_add:
    ws.append(row)
  wb.save(file_path) # Save the changes to the Excel file
  print("\n Your record has been added successfully.")
```

```python
def ContinueAddRecords():
  ch = str(input("Press Y to continue to add record or press any othe character to exit: "))
  if (ch == 'y' or ch =='Y'):
    print("\nAdd Another record\n")
    AddRecords()
    ContinueAddRecords()
  else:
    print("You Exit from Add Records")
    clear_output(wait=True)
```

```python
def UpdateRecords():
  file_path = 'Manager.xlsx'
  wb = load_workbook(file_path)
  ws = wb.active

  cell= str(input('Enter Cell name in that you want to overwrite: '))
  cell_value = ws[cell].value
  print(cell_value)
  new_value = str(input("Enter new value: "))

  # Overwrite the value in the specified cell
  ws[cell] = new_value
   # Save the workbook
  wb.save(file_path)

```

```python
def ContinueUpdateRecords():
  ch = str(input("Press Y to continue to updating record or press any othe character to exit: "))
  if (ch == 'y' or ch =='Y'):
    print("\n Update Another record\n")
    UpdateRecords()
    ContinueUpdateRecords()
  else:
    print("You Exit from Add Records")
    clear_output(wait=True)
```

```python
def SearchRecords(word):
  file_path = 'Manager.xlsx'
  df = pd.read_excel(file_path)

   # Search for the word in the DataFrame
  result = df[df.apply(lambda row: row.astype(str).str.contains(word, case=False).any(), axis=1)]

  if not result.empty:
    print("Search results:")
    print(result.to_string(index=False))  # Print entire rows without index
  else:
    print("No matching results found.")
```

```python
def ContinueSearchRecords():
  ch = str(input("Press Y to continue to Search record or press any othe character to exit: "))
  if (ch == 'y' or ch =='Y'):
    print("\nAdd Another record\n")
    SearchRecords()
    ContinueSearchRecords()
  else:
    print("You Exit from Add Records")
    clear_output(wait=True)
    Manager()
```

```python
def DeleteRecords():
  Id_Number = str(input("Enter ID_number of Employee that to be delet: "))
  file_name = 'Manager.xlsx'
  df = pd.read_excel(file_name)

  result = df[df.apply(lambda row: row.astype(str).str.contains(ID_Number, case=False).any(), axis=1)]
  row_index = df[df.apply(lambda row: row.astype(str).str.contains(ID_Number, case=False).any(), axis=1)].index[0]


  if not result.empty:
    print("\nSearch results:")
    print(result.to_string(index=False))
   # Delete the row
    df.drop(row_index, inplace=True)
    # Save the modified DataFrame back to the Excel file
    df.to_excel(file_name, index=False)
    print("\nRow containing that ID_Number has deleted successfully.")

  else:
   print("No matching results found So Nothind has been deleted.")
```

```python
def ContinueDeleteRecords():
  ch = str(input("Press Y to continue to delete record or press any othe character to exit: "))
  if (ch == 'y' or ch =='Y'):
    print("\nAdd Another record\n")
    DeleteRecords()
    ContinueDeleteRecords()
  else:
    print("You Exit from Add Records")
    clear_output(wait=True)
    Manager()
```
